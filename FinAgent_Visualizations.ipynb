

## Financial Metrics Visualizations

These visualizations compare key financial metrics for AAPL, NVDA, and JPM using results from the FinAgent analysis. The charts show differences in short-term returns and P/E ratios used to support the investment classifications.




```python
import matplotlib.pyplot as plt

# FinAgent comparison results
stocks = ['AAPL', 'NVDA', 'JPM']
returns = [4.7, 4.8, -2.9]

plt.figure(figsize=(8, 5))
bars = plt.bar(stocks, returns)

# Titles and labels
plt.title('3-Month Return Comparison')
plt.xlabel('Stock')
plt.ylabel('3-Month Return (%)')

# Zero line
plt.axhline(0, linewidth=0.8)

# FinAgent favorable threshold
plt.axhline(5, linestyle='--', linewidth=1.2,
            label='Favorable Threshold (5%)')

# Value labels
for bar, value in zip(bars, returns):
    if value >= 0:
        y = value + 0.12
        va = 'bottom'
    else:
        y = value + 0.18
        va = 'bottom'

    plt.text(
        bar.get_x() + bar.get_width()/2,
        y,
        f'{value}%',
        ha='center',
        va=va
    )

plt.ylim(-3.5, 5.8)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# P/E ratios from FinAgent results
stocks = ['AAPL', 'NVDA', 'JPM']
pe_ratios = [39.33, 27.45, 13.99]

plt.figure(figsize=(8, 5))
bars = plt.bar(stocks, pe_ratios)

# Titles and labels
plt.title('P/E Ratio Comparison')
plt.xlabel('Stock')
plt.ylabel('Trailing P/E Ratio')

# FinAgent value threshold
plt.axhline(
    20,
    linestyle='--',
    linewidth=1.2,
    label='Value Threshold (P/E ≤ 20)'
)

# Add values above bars
for bar, value in zip(bars, pe_ratios):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        value + 0.6,
        f'{value:.2f}',
        ha='center',
        va='bottom'
    )

plt.ylim(0, 45)
plt.legend()
plt.tight_layout()
plt.show()
```
